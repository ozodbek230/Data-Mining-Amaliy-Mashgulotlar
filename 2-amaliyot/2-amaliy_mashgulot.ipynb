**Mavzu:** CSV fayllaridagi ma'lumotlar to'plamlarini tahlil qilish, atributlar, umumiy populyatsiya, gipotezalar, diskret/uzluksiz turlari va o'lchov shkalalari.

```python
# Kerakli kutubxonalarni import qilish
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Datasetlar joylashuvi
DATASETS_DIR = 'datasets'
print('Kutubxonalar muvaffaqiyatli yuklandi.')
```

## 1-BO'LIM: 7-topshiriq - Futbol klublari statistikasi
### 1. CSV jadvalining atributlarini chop etish

```python
# 1. CSV faylini o'qish va atributlarni chiqarish
df_football = pd.read_csv(os.path.join(DATASETS_DIR, 'topshiriq_07_futbol_klublar_statistikasi.csv'))
print(f'Jadval hajmi: {df_football.shape[0]} qator x {df_football.shape[1]} ustun\n')
print('Atributlar ro\'yxati va turlari:')
display(df_football.dtypes.to_frame('Dtype'))
```

### 2. Tasodifiy 10 ta obyektni chop etish

```python
# 2. Tasodifiy 10 ta obyekt (Namuna)
sample_football = df_football.sample(n=10, random_state=42)
display(sample_football)
```

### 3 & 4. O'zgaruvchilar va Umumiy Populyatsiya (Bosh to'plam) tavsifi
| Atribut nomi | O'zgaruvchi tavsifi | Umumiy populyatsiya (Bosh to'plam) |
| :--- | :--- | :--- |
| `Jamoa_Nomi` | Klubning rasmiy nomi (unikal toifaviy belgi) | Mamlakatdagi barcha professional futbol klublari |
| `Liga` | Klub o'ynaydigan liga (Superliga, Pro, Birinchi) | Milliy futbol assotsiatsiyasidagi barcha ligalar |
| `Oyinlar_Soni` | Mavsumda o'tkazilgan jami uchrashuvlar | Chempionat nizomidagi barcha o'yinlar miqdori |
| `Galabalar` | Jamoa g'alaba qozongan o'yinlar soni | Barcha o'yinlardagi potensial g'alabalar [0, N] |
| `Duranglar` | Durang qayd etilgan uchrashuvlar soni | Barcha o'yinlardagi potensial duranglar |
| `Maglubiyatlar` | Mag'lubiyatlar soni | Barcha o'yinlardagi potensial mag'lubiyatlar |
| `Ochkolar` | To'plangan jami ochkolar (3*G + 1*D) | Nazariy to'planishi mumkin bo'lgan ochkolar [0, 3*N] |
| `Urilgan_Gollar` | Raqib darvozasiga urilgan to'plar | Barcha o'yinlarda kiritilishi mumkin bo'lgan gollar |
| `Otkazilgan_Gollar` | O'z darvozasidan o'tkazilgan gollar | Barcha o'tkazib yuborilishi mumkin bo'lgan gollar |
| `Gollar_Farqi` | Urilgan va o'tkazilgan gollar farqi | Musobaqadagi barcha nazariy differensiallar [-M, +M] |
| `Xavfli_Vaziyatlar` | Yaratilgan xavfli golli vaziyatlar soni | Chempionatdagi barcha xavfli hujumlar to'plami |
| `Sovrinlar_Oxirgi_5_Yil` | So'nggi 5 yilda yutilgan kuboklar | So'nggi 5 yillikdagi barcha musobaqa sovrinlari |

### 5. Gipotezalar va ularni ma'lumotlar asosida tekshirish

```python
# Gipoteza 1: G'alabalar va Ochkolar o'rtasidagi bog'liqlik
r1 = df_football['Galabalar'].corr(df_football['Ochkolar'])
print(f'H1 tekshiruvi: G\'alabalar va Ochkolar korrelyatsiyasi = {r1:.4f}')

# Gipoteza 2: Xavfli vaziyatlar va Urilgan gollar bog'liqligi
r2 = df_football['Xavfli_Vaziyatlar'].corr(df_football['Urilgan_Gollar'])
print(f'H2 tekshiruvi: Xavfli vaziyatlar va Urilgan gollar korrelyatsiyasi = {r2:.4f}')

# Gipoteza 3: Ligalar bo'yicha o'rtacha ko'rsatkichlar
print('\nH3 tekshiruvi: Ligalar kesimidagi o\'rtacha ko\'rsatkichlar:')
display(df_football.groupby('Liga')[['Urilgan_Gollar', 'Otkazilgan_Gollar', 'Ochkolar']].mean().round(1))
```

### 6 & 7. Diskret/Uzluksiz turlar hamda O'lchov shkalalari
* **Diskret ma'lumotlar:** Jamoa_Nomi, Liga, Oyinlar_Soni, Galabalar, Duranglar, Maglubiyatlar, Ochkolar, Urilgan_Gollar, Otkazilgan_Gollar, Gollar_Farqi, Xavfli_Vaziyatlar, Sovrinlar_Oxirgi_5_Yil.  
* **Uzluksiz ma'lumotlar:** Mavjud emas (barcha qiymatlar butun sonli va toifaviy).  

* **Nominal shkala:** `Jamoa_Nomi` (Nomlar, toifalar).  
* **Tartib (Ordinal) shkala:** `Liga` (Superliga > Pro-liga > Birinchi liga).  
* **Intervalli (Interval) shkala:** `Gollar_Farqi` (Farq hisoblanadi, 0 qiymat tenglikni bildiradi, manfiy qiymatlar mavjud).  
* **Nisbiy (Ratio) shkala:** `Oyinlar_Soni`, `Galabalar`, `Duranglar`, `Maglubiyatlar`, `Ochkolar`, `Urilgan_Gollar`, `Otkazilgan_Gollar`, `Xavfli_Vaziyatlar`, `Sovrinlar_Oxirgi_5_Yil`.

## 2-BO'LIM: 3-topshiriq - Talabalar ta'lim monitoringi va chetlatilish xavfi
### 1. CSV jadvalining atributlarini chop etish

```python
# 1. Talabalar datasetini o'qish va atributlarni chiqarish
df_talaba = pd.read_csv(os.path.join(DATASETS_DIR, 'topshiriq_03_talim_talabalar_xavfi.csv'))
print(f'Jadval hajmi: {df_talaba.shape[0]} qator x {df_talaba.shape[1]} ustun\n')
display(df_talaba.dtypes.to_frame('Dtype'))
```

### 2. Tasodifiy 10 ta obyektni chop etish

```python
# 2. Tasodifiy 10 ta talaba namunasi
display(df_talaba.sample(n=10, random_state=42))
```

### 3 & 4. O'zgaruvchilar va Umumiy Populyatsiya (Bosh to'plam) tavsifi
| Atribut nomi | O'zgaruvchi tavsifi | Umumiy populyatsiya (Bosh to'plam) |
| :--- | :--- | :--- |
| `Talaba_ID` | Talabaning tizimdagi unikal identifikatori | Universitetdagi barcha ro'yxatdan o'tgan talabalar |
| `Fakultet` | Tahsil olayotgan fakultet nomi | OTMdagi barcha mavjud fakultetlar va yo'nalishlar |
| `Kurs` | Talabaning o'quv kursi (1, 2, 3, 4) | Bakalavriatning barcha o'quv kurslari [1-4] |
| `Davomat_Foizi` | Darslarga qatnashish foizi (0-100%) | Semestr davomidagi nazariy davomat oralig'i [0-100%] |
| `Oraliq_Baho_Ortacha` | Oraliq nazorat imtihonlari o'rtacha balli | Baholash tizimidagi barcha oraliq ballar [0-100] |
| `Mustaqil_Ish_Foizi` | Mustaqil ishlarning bajarilish foizi | Barcha belgilangan mustaqil ish topshiriqlari |
| `Ijtimoiy_Yordam` | Ijtimoiy imtiyoz mavjudligi (Ha/Yo'q) | OTMdagi ijtimoiy himoyaga muhtoj barcha talabalar |
| `Yotoqxonada_Turadi` | TTJda istiqomat qilishi (Ha/Yo'q) | Yotoqxona bilan ta'minlangan barcha talabalar |
| `Haftalik_Ishlash_Soati` | Darsdan tashqari mehnat soatlari | Ishlaydigan va ishlamaydigan talabalarning bandlik soatlari |
| `Xavf_Guruhi` | Akademik xavf toifasi (Past, O'rta, Yuqori) | Barcha talabalarning xavf guruhlari bo'yicha to'plami |

### 5. Gipotezalar va ularni ma'lumotlar asosida tekshirish

```python
# Gipoteza 1: Davomat va Mustaqil ish foizi o'rtasidagi bog'liqlik
r_dav = df_talaba['Davomat_Foizi'].corr(df_talaba['Mustaqil_Ish_Foizi'])
print(f'H1 tekshiruvi: Davomat va Mustaqil ish foizi korrelyatsiyasi = {r_dav:.4f}')

# Gipoteza 2: Haftalik ishlash soati va Xavf guruhi
print('\nH2 tekshiruvi: Xavf guruhlari bo\'yicha o\'rtacha ishlash soati:')
display(df_talaba.groupby('Xavf_Guruhi')['Haftalik_Ishlash_Soati'].agg(['mean', 'count']).round(1))

# Gipoteza 3: Davomat ko'rsatkichi va Xavf guruhi
print('\nH3 tekshiruvi: Xavf guruhlari bo\'yicha o\'rtacha davomat foizi:')
display(df_talaba.groupby('Xavf_Guruhi')['Davomat_Foizi'].agg(['mean', 'count']).round(1))
```

### 6 & 7. Diskret/Uzluksiz turlar hamda O'lchov shkalalari
* **Diskret ma'lumotlar:** `Talaba_ID`, `Fakultet`, `Kurs`, `Ijtimoiy_Yordam`, `Yotoqxonada_Turadi`, `Haftalik_Ishlash_Soati`, `Xavf_Guruhi`.  
* **Uzluksiz ma'lumotlar:** `Davomat_Foizi`, `Oraliq_Baho_Ortacha`, `Mustaqil_Ish_Foizi`.  

* **Nominal shkala:** `Talaba_ID`, `Fakultet`, `Ijtimoiy_Yordam`, `Yotoqxonada_Turadi`.  
* **Tartib (Ordinal) shkala:** `Kurs` (1 < 2 < 3 < 4), `Xavf_Guruhi` (Past xavf < O'rta xavf < Yuqori xavf).  
* **Intervalli shkala:** Sof intervalli ustun mavjud emas.  
* **Nisbiy (Ratio) shkala:** `Davomat_Foizi`, `Oraliq_Baho_Ortacha`, `Mustaqil_Ish_Foizi`, `Haftalik_Ishlash_Soati`.
